```python
def calculate_social_impact(dcr, sdg_alignment, people_affected, has_multiplier_effect):
    ME = 1.5 if has_multiplier_effect else 1
    CI = people_affected

# #     SI = dcr * sdg_alignment + CI * ME
#     max_dcr = 500000 # maximum CO2 emission reduction
#     max_sdga = 1 #maximum sdga alignment possible
#     max_ci = 50000 #maximum people affected / jobs created
#     max_me = 1.5 #maximum ME possible with formula

    SI = ((1-1/dcr) * sdg_alignment * CI * ME ) / (max_dcr * max_sdga * max_ci * max_me)

    return SI

# Example
DCR = 1000  # 1000 tons of CO2-equivalent emissions reduced
SDGA = 0.8  # Strong alignment with SDG 13 (Climate Action)
people_affected = 50  # 50 new jobs created
has_multiplier_effect = True  # Project has a cascading impact

social_impact = calculate_social_impact(DCR, SDGA, people_affected, has_multiplier_effect)
print(f"Social Impact: {social_impact}")
```
